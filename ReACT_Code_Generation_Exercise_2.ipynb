```python
# Exercise 2: Code Generation with ReACT Prompting
# This notebook demonstrates a ReACT-style workflow:
# Reason → Generate Code → Run → Observe → Fix

print("=== ReACT Workflow Start ===")

# -----------------------------------
# STEP 1: Reason / Plan
# -----------------------------------

reasoning = """
Problem: Write a Python function to determine if a number is prime.

Plan:
1. Numbers less than 2 are not prime.
2. Check divisibility from 2 up to the square root of the number.
3. If any divisor divides evenly, the number is not prime.
4. Otherwise the number is prime.
"""

print("\nSTEP 1: REASON / PLAN")
print(reasoning)


# -----------------------------------
# STEP 2: Generate Code
# -----------------------------------

print("\nSTEP 2: GENERATED CODE")

import math

def is_prime(n):
    """
    Returns True if n is a prime number.
    Returns False otherwise.
    """

    if n < 2:
        return False

    for i in range(2, int(math.sqrt(n)) + 1):
        if n % i == 0:
            return False

    return True

print("Prime checking function created.")


# -----------------------------------
# STEP 3: Run Code
# -----------------------------------

print("\nSTEP 3: RUN CODE")

test_numbers = [1, 2, 3, 4, 7, 10, 13]

for num in test_numbers:
    print(f"{num} -> {is_prime(num)}")


# -----------------------------------
# STEP 4: Observe Output
# -----------------------------------

print("\nSTEP 4: OBSERVE")

print("Numbers < 2 should return False.")
print("Prime numbers should return True.")
print("Composite numbers should return False.")

print("\nObservation:")
print("The function correctly identifies prime and non-prime numbers.")


# -----------------------------------
# STEP 5: Fix / Improve
# -----------------------------------

print("\nSTEP 5: IMPROVEMENT")

print("""
Improvement:
Instead of checking divisibility up to n,
we check only up to sqrt(n). This reduces
computation and improves performance.
""")

print("\n=== ReACT Workflow Complete ===")
```
