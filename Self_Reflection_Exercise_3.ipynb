```python
# Exercise 3: Self-Reflection Prompt for Improving Output
# Goal: Ask the AI to critique and improve its own summary against explicit requirements.
# This version uses a mock AI (no API) while still demonstrating prompt engineering principles.

import textwrap

def mock_ai_self_reflect(source_text: str, before_summary: str, criteria: dict) -> dict:
    """
    Mock AI that:
    1) critiques the BEFORE summary using explicit criteria
    2) produces an improved AFTER summary that follows those criteria
    """
    critique = []

    # Criteria checks
    length_min, length_max = criteria["length_words"]
    before_words = len(before_summary.split())

    if before_words < length_min or before_words > length_max:
        critique.append(f"- Length: BEFORE is {before_words} words; required {length_min}-{length_max} words.")

    if criteria.get("audience"):
        critique.append(f"- Audience/Tone: Needs to match '{criteria['audience']}' with a professional, neutral tone.")

    if "2 benefits" in criteria.get("content_requirements", ""):
        critique.append("- Content: Must include 2 concrete benefits; BEFORE is vague (no concrete details).")

    if "1 downside" in criteria.get("content_requirements", ""):
        critique.append("- Content: Must include 1 potential downside; BEFORE includes none.")

    if criteria.get("no_hype"):
        critique.append("- Style: Avoid hype/opinion wording; use cautious, factual phrasing aligned with the source.")

    # Improved summary (crafted to satisfy constraints)
    after_summary = (
        "Trials of a four-day workweek have been associated with stable or improved productivity and better employee well-being "
        "in some organizations. Reported benefits include maintaining output while reducing burnout and improving work-life balance. "
        "However, outcomes can vary by industry and role, and implementation may require redesigning workflows, ensuring coverage, "
        "and setting clear performance metrics to avoid service gaps."
    )

    # Ensure AFTER meets word count (adjust if needed)
    after_words = after_summary.split()
    if len(after_words) > length_max:
        after_summary = " ".join(after_words[:length_max])
    elif len(after_words) < length_min:
        after_summary += " " + "Additional evaluation is often needed to confirm results across teams."

    return {
        "critique": "\n".join(critique),
        "after_summary": after_summary
    }


# -------------------------------
# 1) Source text (what we summarize)
# -------------------------------
source_text = """
Some companies trialing a four-day workweek report stable or improved productivity, alongside improvements in employee well-being.
However, results vary by industry and role. Implementation often requires redesigning workflows, managing coverage, and setting clear performance metrics.
"""

# -------------------------------
# 2) BEFORE summary (intentionally weak)
# -------------------------------
before_summary = "The four-day workweek is good because it improves productivity and happiness. Companies should adopt it."

# -------------------------------
# 3) Self-reflection prompt criteria
# -------------------------------
criteria = {
    "accuracy": "Must match claims in the source text; do not add unsupported facts.",
    "clarity": "Use specific, concrete phrasing; avoid vague statements.",
    "audience": "BUS4 118S student (professional and neutral tone)",
    "length_words": (70, 90),
    "formatting": "Single paragraph summary.",
    "content_requirements": "Include 2 benefits and 1 downside based on the source text.",
    "no_hype": True
}

# -------------------------------
# 4) Self-reflection prompt (what we'd send to an AI)
# -------------------------------
self_reflection_prompt = f"""
Source text:
{source_text.strip()}

BEFORE summary:
{before_summary}

Critique requirements:
- Accuracy vs source text (no unsupported claims)
- Clarity and specificity (avoid vague wording)
- Audience: {criteria['audience']}
- Length: {criteria['length_words'][0]}–{criteria['length_words'][1]} words
- Format: {criteria['formatting']}
- Content: {criteria['content_requirements']}
- Style: Avoid hype/opinion wording; keep a neutral tone

Task:
1) Provide a critique of the BEFORE summary using the requirements above.
2) Then write an improved AFTER summary that follows all requirements.
Output format:
Critique:
- bullets

After summary:
<single paragraph>
"""

# -------------------------------
# 5) Run mock self-reflection
# -------------------------------
result = mock_ai_self_reflect(source_text, before_summary, criteria)

print("=== EXERCISE 3: SELF-REFLECTION PROMPT ===\n")

print("PROMPT USED (Self-Reflection Prompt):\n")
print(textwrap.dedent(self_reflection_prompt).strip())

print("\n\n--- BEFORE SUMMARY ---")
print(before_summary)

print("\n\n--- CRITIQUE ---")
print(result["critique"])

print("\n\n--- AFTER SUMMARY (Improved) ---")
print(result["after_summary"])

print("\n\n=== END ===")
```
